## Electrostatic Environment

A point charge or charge distribution creates both an electric field ($\vec{E}=kQ/r^2 \vec{r}$ or *insert integral form*) and electric potential ($V=kQ/r$ or *insert integral form*). The Lichtenberg process starts with the introduction of high *electric potential (voltage)* on a dielectric material. Electric potential, in this case, is created by a *distribution of charge $(\rho)$* and is expressed through the use of the Differential form of Gauss’s law [<a href = "http://hyperphysics.phy-astr.gsu.edu/hbase/electric/diverg.html#c1">1</a>], 

**Electrostatic environment equation (General case):**

<div align="center">

${\nabla}{\cdot}{E} = \frac {\rho}{\epsilon_0}$, where \
\
${E}$ = electric field \
${\rho}$ = charge density\
${\epsilon_0}$ = permittivity
</div>

The base environment electrostatic environment $(\rho = 0)$ follows the principles Laplace’s Equation, 
<div align="center">

${\nabla}^2{V} = 0$ 

</div>

wherein the solutions to these equations are known as *harmonic functions*. These harmonic functions are twice continuously differentiable [<a href = "https://ocw.mit.edu/courses/18-04-complex-variables-with-applications-spring-2018/2e739bb156efb0bc7103fc43d0897dda_MIT18_04S18_topic5.pdf">2</a>]. Since the first derivative measures the slope and the second derivative measures the change in slope and continuity, harmonic function provides the average value of all the surrounding values at a point. Due to this averaging, there are no local voltage peaks or valleys inside the material, but concentrated on the boundaries of the system.

If the Laplace’s Equation acts as the general landscape, the *Potential Gradient* is what dictates the direction of the Lichtenberg figures propagation, given by the equation: 
<div align = "center"> 

$(E=-\nabla{V})$ 

</div>

This sets up the site where Ionization will happen. This happens when the Potential Gradient becomes massive (Note: process to be discussed on Field Enhancement) enough that:
<div align = "center"> 

$(|\nabla{V}|\ge E_{crit}$, where \
\
$E_{crit}$ = Dielectric strength or the maximum electric field that an insulator can handle before it breaks down and becomes a conductor
</div>

When this happens, the electric field is strong enough to rip an electron from the atom. At the moment of breakdown, there is now a space charge in the interior of the material. Therefore, the new electrostatic environment could be better described as ${\nabla}^2{V} = \frac {\rho}{\epsilon_0}$

The ionized point, being conductive, is now acting as a dynamic moving boundary for the next electric field calculation until it stops when $|E| < E_{crit}$
