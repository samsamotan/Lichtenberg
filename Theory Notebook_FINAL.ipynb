# Lichtenberg
Numerical simulation of dielectric breakdown generating Lichtenberg figures in Acrylic and Wood

## Lichtenberg Figures

### Summary
A Lichtenberg Figure is a branching fractal pattern that is formed when a high electric potential is applied to a dielectric (insulating) material, where the potential distribution follows Laplace’s Equation ($\nabla^2 V = 0$) until a high potential gradient triggers localized breakdown. When the local field exceeds the material’s dielectric strength, breakdown occurs, introducing free electrons into the system and shifting the electrostatic environment according to the behavior of Poisson’s equation ${\nabla}^2{V} = \frac {\rho}{\epsilon_0}$. 

Breakdown is driven by avalanche ionization, where the applied electric field accelerates free electrons, making them collide with lattice atoms. This creates a cascading chain reaction of self-multiplying electrons that leads to the breakage of chemical bonds and the formation of conductive channels. 

The branching structure arises from field enhancement, where electric fields intensify at sharp tips or irregularities. The plasma channels that occur from the resulting impact ionization create a self-propagating growth mechanism, as these channels ensure that the localized potential gradient is steep enough for further ionization. 

Despite being governed by deterministic equations, the growth process is influenced by randomness, such as material defects and stochastic ionization, resulting in fractal geometry. The resulting Lichtenberg Figures exhibit self-similarity, characterized by a fractal dimension between 1 < D < 2. The processes creating the Lichtenberg figure support this branching property through Field Screening and Laplacian Shielding. 

### Descriptive Visualization

A Lichtenberg Figure is a branching electric discharge that has self-similar patterns across different scales, exhibiting fractal properties. The shape of the Lichtenberg Figure depends on the polarity of the electric charge that is applied to the medium. A positive charge (applied by an anode) would create a Lichtenberg with a widely extending patch, composed of a dense nucleus with branches that radiate in all directions. In contrast, a negative charge (applied by a cathode) would produce a figure that is smaller with a fan-like boundary.

### Material Consideration

In practice, Lichtenberg figures can be observed in different material setups. 

Case 1: Acrylic 

Acrylic is a homogenous dielectric, which means its internal structure is uniform. Lichtenberg Figures in acrylic are created by injecting high-energy electrons using a linear accelerator (LINAC). 

Case 2: Wood 

Wood is a heterogeneous material with natural variations in grain and composition. For this setup, two electrodes are placed on the surface, and a high voltage is applied across them. 


## Electrostatic Environment

A point charge or charge distribution creates both an electric field ($\vec{E}=kQ/r^2 \vec{r}$ or *insert integral form*) and electric potential ($V=kQ/r$ or *insert integral form*). The Lichtenberg process starts with the introduction of high *electric potential (voltage)* on a dielectric material. Electric potential, in this case, is created by a *distribution of charge $(\rho)$* and is expressed through the use of the Differential form of Gauss’s law [<a href = "http://hyperphysics.phy-astr.gsu.edu/hbase/electric/diverg.html#c1">1</a>], 

**Electrostatic environment equation (General case):**

<div align="center">

${\nabla}{\cdot}{E} = \frac {\rho}{\epsilon_0}$ (Poisson's Equation), where \
\
${E}$ = electric field \
${\rho}$ = charge density\
${\epsilon_0}$ = permittivity
</div>

The base environment electrostatic environment $(\rho = 0)$ follows the principles Laplace’s Equation, 
<div align="center">

${\nabla}^2{V} = 0$ 

</div>

wherein the solutions to these equations are known as *harmonic functions*. These harmonic functions are twice continuously differentiable [<a href = "https://ocw.mit.edu/courses/18-04-complex-variables-with-applications-spring-2018/2e739bb156efb0bc7103fc43d0897dda_MIT18_04S18_topic5.pdf">2</a>]. Since the first derivative measures the slope and the second derivative measures the change in slope and continuity, harmonic function provides the average value of all the surrounding values at a point. Due to this averaging, there are no local voltage peaks or valleys inside the material, but concentrated on the boundaries of the system.

If the Laplace’s Equation acts as the general landscape, the *Potential Gradient* is what dictates the direction of the Lichtenberg figures propagation, given by the equation: 
<div align = "center"> 

$(E=-\nabla{V})$ 

</div>

This sets up the site where Ionization will happen. This happens when the Potential Gradient becomes massive (Note: process to be discussed on Field Enhancement) enough that:
<div align = "center"> 

$(|\nabla{V}|\ge E_{crit}$, where \
\
$E_{crit}$ = Dielectric strength or the maximum electric field that an insulator can handle before it breaks down and becomes a conductor
</div>

When this happens, the electric field is strong enough to rip an electron from the atom. At the moment of breakdown, there is now a space charge in the interior of the material. Therefore, the new electrostatic environment could be better described as ${\nabla}^2{V} = \frac {\rho}{\epsilon_0}$

The ionized point, being conductive, is now acting as a dynamic moving boundary for the next electric field calculation until it stops when $|E| < E_{crit}$



## Dielectric Breakdown

**Conductors vs. Dielectrics**

The difference between dielectrics (insulators) and conductors lies in the mobility of electrons and their response to an electric field. In conductors, electrons move freely and are not bound to specific atoms — when an electric field is applied, these free electrons immediately respond by flowing through the material as electric current. In contrast, dielectrics have electrons that are tightly bound to their parent atoms or molecules. In their normal state, dielectrics have very few charge carriers and will conduct very little current with a given electric field. However, with a large enough electric field, the number of charge carriers rapidly increases, the electrical resistance drops, and the material becomes a conductor — this is **Dielectric Breakdown**.

**Dielectric Breakdown** is the sudden and drastic increase in electrical conductivity through an insulating material when the applied electric field exceeds the *Dielectric Strength* — the maximum electric field intensity that a dielectric material can withstand before breakdown. Mathematically, the dielectric strength $E_{bd}$ can be expressed as:

<div align="center">

$E_{bd} = \frac{V_{bd}}{d}$, where \
\
$V_{bd}$ = breakdown voltage (minimum voltage that causes dielectric breakdown to occur) \
$d$ = material thickness

</div>

Once the applied voltage reaches or exceeds $V_{bd}$, free electrons in the material gain sufficient energy from the electric field to initiate **Avalanche Ionization**.

Fundamentally, conductors do not experience dielectric breakdown because they already possess abundant free electrons that move readily under any electric field. However, conductors can still experience other forms of failure when an extraordinarily intense electric field overcomes the nuclear binding forces holding atoms together (atomic disintegration).

**Avalanche Ionization in Solids**

Electron avalanche in a solid is a self-multiplying cascade process where free electrons, accelerated by an electric field, collide with lattice atoms and free additional electrons through impact ionization. Wang et al. (2025) describe this in four sequential stages:

**1. Collision Ionization between electrons and lattice atoms**

Free electrons, driven by the applied electric field $(E_{app})$, undergo inelastic collisions with lattice atoms. When a free electron's kinetic energy exceeds the ionization energy of lattice atoms, the bound electron is knocked into a more excited state (but remains electrostatically attracted to the atom). The ionization rate $\alpha(E_{app})$ describing the dynamic probability of ionization is:

<div align="center">

$\alpha(E_{app}) = \frac{W_{gain}}{W_i} = \frac{q \lambda E_{app}}{W_i}$, where \
\
$\alpha(E_{app})$ = ionization rate \
$q$ = electron charge \
$\lambda$ = electron mean free path \
$W_{gain} = q\lambda E_{app}$ = energy gained by free electron from the applied field \
$W_i$ = ionization energy

</div>

**2. Multiplication of Local Electrons**

The dynamic evolution of the local electron density over the electric field is expressed as:

<div align="center">

$N(r, t) = N_0 \exp\left\{\int_0^t \alpha[E_{app}(t')]\, dt'\right\}$, where \
\
$N_0$ = initial outer electron density

</div>

Due to continuous collisions, excited electrons accumulate around the atom, building up the local electric field $E_{local}$, which can be obtained by solving Poisson's equation:

<div align="center">

$\nabla \cdot (\epsilon_0 \epsilon_r E_{local}(r,t)) = N(r,t)$, where \
\
$\epsilon_0$ = vacuum permittivity \
$\epsilon_r$ = relative permittivity

</div>

**3. Breakage of Chemical Bonds**

The accumulation of $E_{local}$ raises the electric energy of excited electrons $W_{elec}$. When $W_{elec}$ exceeds the bond energy, the chemical bond breaks and electrons are released, continuing to break nearby bonds in a chain reaction. The electric energy is:

<div align="center">

$W_{elec} = \int_V \frac{1}{2} \epsilon_0 \epsilon_r E_{local}^2(r,t)\, dV$

</div>

The bond-breaking probability from position $r$ to adjacent position $r'$ is:

<div align="center">

$p(r \to r') = \frac{(W_{elec}^{r'} / W_b)^\xi}{\sum (W_{elec}^{r'} / W_b)^\xi}$, where \
\
$W_{elec}^{r'}$ = electric energy of excited electrons at position $r'$ \
$W_b$ = bond energy \
$\xi$ = fractal dimension determining avalanche path shape (set to 1 for a linear relationship)

</div>

**4. Formation of a Continuous Avalanche Channel**

As the bond-breaking chain reaction proceeds, the localized density of broken bonds increases, forming a conductive percolation channel. The avalanche is considered complete when the longitudinal channel length reaches ~6 nm (comparable to the electron mean free path in conductive metals). At this point, the dielectric has become a conductor and current is no longer limited by the material's insulating properties. The rate equation for the avalanche channel evolution is:

<div align="center">

$\frac{\partial \eta(r,t)}{\partial t} = A\eta^2 - B\eta + C$, where \
\
$\eta(r,t)$ = normalized variable describing avalanche channel evolution \
$A, B$ = coefficients determining the difficulty and tendency of bond breaking \
$C$ = bond-breaking probability $p$

</div>

## Field Enhancement

In electrostatic equilibrium, the surface charge density $(\sigma)$ and electric field at a point are inversely proportional to the radius of curvature $(r)$ at that point:

<div align="center">

$\frac{\sigma_1}{\sigma_2} = \frac{r_2}{r_1}$ $\quad$ and $\quad$ $\frac{E_1}{E_2} = \frac{r_2}{r_1}$ $\quad$

</div>

In real physical objects, the slightest defect (small $r$) on the surface provides the ideal condition for a massive local electric field due to this inverse proportionality with $r$. In other words, charge concentrates at regions with sharp tips or edges, creating intense local fields. This increase is quantified by $\beta$, the **Field Enhancement Factor** [<a href="https://www.sciencedirect.com/topics/engineering/field-enhancement">5</a>]:

<div align="center">

$\beta = \frac{E_{local}}{E_{avg}}$, where \
\
$E_{local}$ = local electric field at the tip \
$E_{avg}$ = average (background) electric field

</div>

When $E_{local} \geq E_{crit}$, impact ionization occurs and the material changes its localized physical state to **plasma** — a state with free-floating negative electrons and positive ions. This plasma channel now acts similarly to a conductor, wherein resistivity drops and it becomes a voltage extension of the source, following the principle of the *dynamic moving boundary*.

Since this newly created channel tip is sharp, field enhancement repeats at the new tip — creating a **self-propagating growth** that gives rise to the branching structure of Lichtenberg figures.

## Fractal Geometry

The resulting Lichtenberg figure and its self-propagating growth creates a complex and irregular pattern that exhibits self-similarity at various scales of magnification. This behavior follows the idea of fractal geometry.

In a perfectly uniform material, electricity would travel in a straight line. However, despite the deterministic rules set by Laplace and Poisson equations, the growth at each step is stochastic in nature because of local fluctuations and defects. In addition to this, the bond-breaking probability $p(r \to r')$, adds a layer of complexity and randomness in the picture.

**Fractal Dimension ($D$)**

This complexity is better expressed through the concept of Fractal Dimensions. Fractal Dimension is a measure of a pattern’s complexity that quantifies the amount of space it fills as it is zoomed in on a smaller scale. In standard geometry, Euclidean dimensions are expressed in integers (e.g. 1D is a line, 2D is a square, and 3D is a cube), wherein if a dimension is doubled or multiplied by a factor of a, there would be a^n copies of the original where n is the number of dimensions. In contrast, Fractal dimension is expressed as a fraction between these integer dimensions, meaning that it does not follow the same scaling factor of standard geometry. The Lichtenberg figures is more complex than a 1D line but does not completely fill a 2D area, making its dimension $1 < D < 2$. In literature, it is between 1.5 to 2 in two-dimensional surface discharge and between 2 to 2.5 in three-dimensional volume discharge.

**Branching of Fractals**

Another property of these geometric figures is that the branching primarily happens on the “outside” or at the tip. The processes creating the Lichtenberg figure supports this branching property through Field Screening and Laplacian Shielding. Field Screening is the dampening of electric fields caused by the mobile electric charge carriers. On the other hand, Laplacian Shielding happens because the branch is equipotential and \nabla{V} inside the figure drops close to 0. This forces the growth to the perimeter or the tips, ensuring that the figure expands outward.

**Box Counting**

Due to the complex nature of fractals, the Box Counting Method is an often used algorithm to Calculate $D$ from the simulation results. It works in this order:

1. *Grid partitioning*: The simulation domain is overlaid with a grid of square boxes with side length $\epsilon$
2. *Occupied Box Count*: Count the number of boxes, $N(\epsilon)$ with at least one "ionized" pixel
3. *Scaling*: Repeat for different box sizes
4. *Regression*: Plot $\log(N(\epsilon))$ against $\log(1/\epsilon)$

Mathematically, $D$ is the slope of the log-log plot
<div align="center">

$D = \lim\limits_{\epsilon \to 0} \frac{\log(N(\epsilon))}{\log(1/\epsilon)}$

</div>

This equation can be used in sanity checks when comparing simulation values and literature values

## Computational Dielectric Breakdown Model (Theory to Code)

**Dielectric Breakdown in Acrylic**
<br>
In simulating dielectric breakdown in acrylic to form a Lichtenberg figure, we perform the following steps:

1. Define the bounds of the 3-dimensional acrylic block that will serve as our medium. Afterwards, we use numpy's meshgrid() function to give every cell in our block, an (x,y,z) coordinate.
2. When the acrylic block is placed under a linear accelerator (LINAC), electrons are accelerated to relativistic speeds (near light speed) and penetrate the acrylic, where they stop somewhere inside the acrylic block, forming an internal net negative charge (i.e., a space charge). To simulate the electrons inside the acrylic, a Gaussian is used to define a normal charge distribution at a defined depth.
3. Given the presence of a charge density inside the acrylic, we use Poisson's equation to solve the electric potential (voltage), at every (x,y,z) coordinate such that it is very negative the closer it is to the "charge cloud"
4. In Lichtenberg figures in acrylic, the grounded nail puncturing into the acrylic becomes the first cell of the "tree". In simulating this, we set the coordinates where the nail is placed. From there, the 26 neighbors of that cell (6 faces + 8 corners + 12 edges) serve as "candidates" that could break next.
5. From these candidates, we calculate each of their electric fields by dividing the voltage of that edge by its distance from the nearest tree cell; in the first iteration, that would be the nail itself. 
6. Each candidate's electric field is raised to the power η (eta), which controls how "tip-seeking" the discharge is. For instance, a candidate with twice the field strength gets 2^η times the weight. Afterwards, these weighted values are then normalized into a probability distribution that defines the probability of each cell to be the next one that gets broken down.
7. One candidate cell is randomly selected according to this probability distribution. This selected cell "breaks down" and joins the tree, becoming a new grounded node (i.e., V = 0).
8. This new candidate cell now gets its own 26 neighbors as new candidates, from which the loop would continue for thousands of iterations.


**Dielectric Breakdown in Wood**
<br>
In simulating dielectric breakdown in wood to form a Lichtenberg figure, we perform the following steps:

1. Unlike the simulation in acrylic, Lichtenberg formation on wood occurs on its surface only as the discharge travels across its surface. Therefore, we define the bounds of a 2-dimensional rectangular grid that represents the flat surface of the wood board. Furthermore, in each of those (x,y) cells, we store a temporary 0 for the electric potential at each cell. 
2. In real Lichtenberg figure formation on wood, an electrolytic solution is poured onto the surface to decrease the resistance of the wood so that it can facilitate the flow of electric currents on its surface. This is performed by adding a cathode and an anode on either ends of the board, resulting in a very high potential difference between the two electrodes. Specifically, the cathode is defined to have a voltage of -10V while the anode is to have a voltage of +10V.
3. Note that since there are no charges trapped inside unlike the acrylic simulation, we can now just solve the Laplace equation instead of the Poisson equation. Doing so gives us the gradient of electric potential (voltages) across the 2D grid of the wooden board.
4. There are different ways to start the Lichtenberg figure formation, whether that be single-tree or double-tree. In the case of single-tree, where the Lichtenberg figure only forms from one side, it is simulated by defining our starting point on the cathode. Like the acrylic simulation, this cell now becomes the start of our "tree". Afterwards, we define the "candidates" where it could proceed next as either the 4 cells (up, down, left, right) or the 8 cells (includes the diagonals) that are the neighbors of that cell.
5. Importantly, wood has a property called **anisotropy** where the grain of the wood itself becomes a variable in determining the path that dielectric breakdown will follow. To simulate this, we define a variable g that is stored inside each cell of the wooden board that shows the conductivity for a certain cell (e.g., more dense areas along grain are more conductive, and are given a higher value for g) in such a way that it mimics the grain pattern of wood (e.g., horizontal strips, or random homogeneity). Then, in computing for the weights and probabilities, this factor g is also multiplied to it.
6. Similar to the acrylic simulation, we calculate for the weights of each candidate such that the higher the voltage of a candidate (the less negative it is), the higher it's weight is, and thus, the higher the probability that is defined for that candidate.
7. We choose a random candidate given the different probabilities from the candidates. After which, either the next 4 or 8 neighbors of the chosen cell become the new candidates.
8. In the event of double-tree branching, two starting points (i.e., two different trees) from the opposite electrodes are defined such that both are moving towards the center. The simulation then stops once a cell from one tree is adjacent to a cell from the other tree.
