# Lichtenberg Figures: Theoretical Foundations

Lichtenberg figures are branching, dendritic discharge patterns produced when electrical energy is dissipated through or across an insulating medium. First documented by Georg Christoph Lichtenberg in 1777, they share a common fractal morphology rooted in the stochastic, field-driven nature of electrical discharge.

This notebook develops the theoretical framework in three parts:

1. **Common theory** — the electrostatics, fractal geometry, and discharge physics shared by all Lichtenberg figure types
2. **Charge-implanted dielectric figures** — bulk breakdown in electron-irradiated PMMA (the "acrylic" method)
3. **Resistive surface burning** — current-driven carbonization on wood (the "wood burning" method)

---
# Part I — Common Theory
---

## 1. Fractal Geometry and Statistical Self-Similarity

All Lichtenberg figures — regardless of formation method — exhibit branching morphologies that are statistically self-similar across multiple length scales. This classifies them as natural fractals and provides a unified quantitative language for comparing figures produced by very different physical processes.

### 1.1 Hausdorff (Box-Counting) Dimension

The fractal dimension $D_f$ is operationally defined via the box-counting algorithm. The figure is covered with a grid of cells of side length $\epsilon$, and the number of occupied cells $N(\epsilon)$ is recorded.

**Derivation.** For an ordinary geometric object of integer dimension $d$, scaling the measurement resolution by a factor $\lambda$ changes the number of required covering elements as $N \propto \lambda^d$. A line ($d=1$) cut into pieces of length $\epsilon$ requires $N = L/\epsilon \propto \epsilon^{-1}$ pieces. A filled square ($d=2$) covered by boxes of side $\epsilon$ requires $N = A/\epsilon^2 \propto \epsilon^{-2}$. In general:

$$N(\epsilon) \propto \epsilon^{-d}$$

Fractals generalize this to non-integer $d \to D_f$:

$$N(\epsilon) \sim \epsilon^{-D_f}$$

Taking the logarithm of both sides:

$$\ln N(\epsilon) = -D_f \ln \epsilon + \text{const}$$

Solving for $D_f$:

$$D_f = -\frac{\ln N(\epsilon)}{\ln \epsilon} \quad \xrightarrow{\text{exact}} \quad D_f = -\lim_{\epsilon \to 0} \frac{\ln N(\epsilon)}{\ln \epsilon}$$

In practice, $D_f$ is obtained as the slope of a linear regression on the $\ln N$ vs. $\ln(1/\epsilon)$ plot over the scaling regime (bounded below by pixel resolution and above by the figure extent).

Reference values:
- Euclidean line: $D_f = 1$
- Space-filling plane: $D_f = 2$
- Charge-implanted acrylic figures (2D projection): $D_f \approx 1.65$–$1.75$
- Wood-burned surface figures: $D_f \approx 1.3$–$1.6$ (lower due to grain-directed anisotropy)
- DLA clusters (theoretical): $D_f \approx 1.71 \pm 0.03$

The non-integer dimensionality reflects that the discharge tree is more space-filling than a smooth curve yet fails to tile the plane.

### 1.2 Connection to Diffusion-Limited Aggregation

The morphological similarity between Lichtenberg figures and Diffusion-Limited Aggregation (DLA) clusters arises from a shared mathematical structure. In DLA, particles executing Brownian motion (governed by the diffusion equation $\partial_t c = D\nabla^2 c$) aggregate irreversibly upon contact with a growing cluster. In the steady state, the concentration field satisfies Laplace's equation $\nabla^2 c = 0$ — identical in form to the electrostatic potential outside a discharge tree.

This correspondence places ideal dielectric breakdown and DLA in the same **universality class** of Laplacian growth phenomena. Wood-burned figures deviate from pure DLA due to the strong anisotropy and heterogeneity of the substrate, but the underlying growth instability — preferential advancement at tips where the field (or current density) is concentrated — is shared.

### 1.3 Lacunarity and Multifractal Analysis

While $D_f$ captures the global scaling behavior, it does not fully characterize spatial heterogeneity. **Lacunarity** $\Lambda(\epsilon)$ quantifies the textural variation at scale $\epsilon$, distinguishing structures with identical $D_f$ but different spatial distributions of branches. **Multifractal analysis** extends this by computing a continuous spectrum of generalized Rényi dimensions $D_q$, where $D_0$ is the box-counting dimension, $D_1$ the information dimension, and $D_2$ the correlation dimension. A broad $D_q$ spectrum indicates strong spatial heterogeneity in the growth probability distribution — typically broader for anisotropic wood figures than for acrylic figures grown in homogeneous media.

## 2. Electrostatics and Field Theory

Both formation methods are governed by electric fields, though they enter in different ways — as the driving force for bulk dielectric failure (acrylic) or as the source of current through a resistive surface layer (wood).

### 2.1 Governing Field Equations

**Derivation of Poisson's equation.** Begin with the two relevant Maxwell equations in a linear, isotropic dielectric:

$$\nabla \cdot \vec{D} = \rho_{\text{free}}$$

where $\vec{D} = \epsilon_0 \epsilon_r \vec{E}$ is the displacement field and $\rho_{\text{free}}$ is the free charge density. For an electrostatic field, $\vec{E} = -\nabla\phi$. Substituting:

$$\nabla \cdot (\epsilon_0 \epsilon_r \vec{E}) = \rho_{\text{free}}$$

For a medium with spatially uniform $\epsilon_r$, this becomes:

$$\epsilon_0 \epsilon_r \nabla \cdot \vec{E} = \rho_{\text{free}}$$

$$\epsilon_0 \epsilon_r \nabla \cdot (-\nabla\phi) = \rho_{\text{free}}$$

$$\boxed{\nabla^2 \phi = -\frac{\rho_{\text{free}}}{\epsilon_0 \epsilon_r}}$$

This is **Poisson's equation**. In charge-free regions ($\rho_{\text{free}} = 0$), it reduces to **Laplace's equation**: $\nabla^2 \phi = 0$.

**Derivation of Ohm's law (microscopic).** In a conducting medium, free charge carriers (electrons or ions) experience an electric force $\vec{F} = q\vec{E}$ and reach a steady-state drift velocity $\vec{v}_d$ determined by the balance of electric driving force and collisional drag:

$$q\vec{E} = \frac{m\vec{v}_d}{\tau_c}$$

where $\tau_c$ is the mean time between collisions and $m$ the carrier mass. The current density is $\vec{J} = nq\vec{v}_d$, where $n$ is the carrier number density. Solving for $\vec{v}_d$ and substituting:

$$\vec{J} = nq \cdot \frac{q\tau_c}{m}\vec{E} = \frac{nq^2\tau_c}{m}\vec{E}$$

Defining the conductivity $\sigma_{\text{cond}} \equiv nq^2\tau_c / m$:

$$\boxed{\vec{J} = \sigma_{\text{cond}} \vec{E} = -\sigma_{\text{cond}} \nabla \phi}$$

In the acrylic case, $\sigma_{\text{cond}} \approx 0$ everywhere until breakdown occurs. In the wood case, $\sigma_{\text{cond}}$ is finite, spatially heterogeneous, and evolves dynamically as carbonization progresses.

### 2.2 Dielectric Breakdown Criterion

Dielectric breakdown is the irreversible transition of an insulator to a conducting state when the local electric field exceeds the **intrinsic breakdown strength** $E_b$:

$$E \geq E_b \quad \Longrightarrow \quad \text{breakdown}$$

**Derivation of the uniform-field relation.** For a parallel-plate geometry with voltage $V_b$ applied across a gap of thickness $d$, the potential varies linearly: $\phi(z) = V_b(1 - z/d)$. The field magnitude is:

$$E = -\frac{d\phi}{dz} = \frac{V_b}{d}$$

At breakdown, $E = E_b$, giving: $\boxed{E_b = V_b / d}$.

Several mechanisms can drive failure depending on the field regime and timescale:

- **Intrinsic (electronic) breakdown**: Impact ionization produces an electron avalanche. Dominant at high fields and short timescales ($< \mu$s). This is the primary mechanism in charge-implanted acrylic.
- **Thermal breakdown**: Joule heating reduces resistivity, creating positive feedback (thermal runaway). Dominant for sustained current flow. This is the primary mechanism in wood burning.
- **Electromechanical breakdown**: Coulombic compression exceeds the material's mechanical yield strength.
- **Partial discharge / treeing**: Progressive erosion by localized discharges at sub-critical bulk fields.

### 2.3 Field Enhancement at Conductor Tips

Both methods involve conductor tips (nails, electrodes) in contact with the medium. On any conductor at fixed potential, mobile surface charges redistribute to maintain an equipotential surface.

**Derivation for a conducting sphere.** Consider an isolated conducting sphere of radius $R$ carrying total charge $Q$. By Gauss's law, the field outside the sphere is identical to that of a point charge:

$$\oint \vec{E} \cdot d\vec{A} = \frac{Q}{\epsilon_0} \quad \Longrightarrow \quad E(r) \cdot 4\pi r^2 = \frac{Q}{\epsilon_0} \quad \Longrightarrow \quad E(r) = \frac{Q}{4\pi\epsilon_0 r^2}$$

At the surface ($r = R$): $E_{\text{surface}} = Q / (4\pi\epsilon_0 R^2)$.

The potential of the sphere is $V = Q / (4\pi\epsilon_0 R)$, so $Q = 4\pi\epsilon_0 R V$. Substituting:

$$E_{\text{surface}} = \frac{4\pi\epsilon_0 R V}{4\pi\epsilon_0 R^2} = \boxed{\frac{V}{R}}$$

Halving the radius doubles the field. A needle tip is the limiting case: $r_{\text{tip}}$ is extremely small ($\sim$10–100 $\mu$m), concentrating the field into a microscopic region.

**General tip geometry.** For a hyperboloidal tip of apex radius $r_{\text{tip}}$ opposite a distant plane at separation $d$, solving Laplace's equation in prolate spheroidal coordinates yields:

$$E_{\text{tip}} \sim \frac{V}{r_{\text{tip}} \ln(2d/r_{\text{tip}})}$$

Enhancement factors $\beta = E_{\text{tip}}/E_{\text{avg}}$ of $10^2$–$10^3$ are typical for $r_{\text{tip}} \sim 100\;\mu\text{m}$. This is relevant to both methods: in acrylic, it controls where breakdown nucleates; in wood, it determines where the initial current concentration and carbonization begin.

## 3. Discharge Propagation and Branching Physics

Once a conducting channel is established — whether by dielectric breakdown in a solid or by carbonization on a surface — it must propagate and branch to produce the characteristic Lichtenberg morphology. The physics of propagation differs between the two methods, but the branching instability has a common origin.

### 3.1 The Tip-Splitting Instability

In both methods, the advancing front of the discharge (or burn) concentrates the driving field at its tips. A smooth, planar front is unstable to perturbations: any protrusion extends into a region of higher field (or current density), causing it to advance faster, which further enhances the field at that point. This is the **Mullins-Sekerka instability**, the same mechanism responsible for dendritic solidification in crystal growth.

The branching pattern is then shaped by two competing effects:
- **Field concentration at tips** drives growth preferentially at protrusions (destabilizing)
- **Material properties** (homogeneity, isotropy, thermal diffusion) tend to smooth the front (stabilizing)

When the destabilizing effect dominates — as it does in both high-field dielectric breakdown and localized resistive heating — the result is a fractal branching tree.

### 3.2 Polarity-Dependent Morphology

For methods involving gaseous or solid dielectric breakdown (primarily the acrylic method, but also relevant to surface arcing on wood), the discharge polarity affects morphology:

**Positive (cathode-directed) streamers** propagate in the direction of electron drift. The streamer head carries a net positive space charge that enhances the field ahead of the front, seeding secondary avalanches via photoionization. This produces **densely branched, radially symmetric structures** with high $D_f$.

**Negative (anode-directed) streamers** propagate against electron drift. The advancing front must generate seed electrons through field ionization alone, requiring higher fields for sustained propagation. This yields **sparser, less branched structures** with lower $D_f$.

This polarity asymmetry is a direct consequence of non-equilibrium ionization kinetics and is most pronounced in the acrylic method. In wood burning, the distinction is less sharp because the dominant mechanism is thermal (Joule heating) rather than ionization-driven, though some polarity dependence persists due to surface arcing near the electrodes.

# Part II — Charge-Implanted Dielectric Figures (Acrylic Method)
---

## 4. Formation Process

### 4.1 Charge Implantation

High-energy electrons (typically 5–25 MeV) from a linear accelerator are implanted into a PMMA substrate. The electron penetration depth is governed by the Bethe-Bloch stopping power of the material and scales approximately with beam energy. The implanted electrons thermalize and become immobilized within the bulk, forming a quasi-planar space charge layer at a characteristic depth $z_0$.

### 4.2 Space Charge Accumulation

Due to the negligible conductivity of PMMA ($\rho > 10^{15}\;\Omega\cdot\text{cm}$), the implanted charge is effectively frozen in place.

**Derivation of internal field from surface charge density.** Model the implanted charge as an infinite plane of surface charge density $\sigma$ embedded in a dielectric of permittivity $\epsilon_r$. Apply Gauss's law to a cylindrical Gaussian surface of cross-sectional area $A$ straddling the charge plane:

$$\oint \vec{D} \cdot d\vec{A} = Q_{\text{enc}} = \sigma A$$

By symmetry, $\vec{D}$ is perpendicular to the plane and equal on both sides:

$$2 D \cdot A = \sigma A \quad \Longrightarrow \quad D = \frac{\sigma}{2}$$

For the field on one side of the plane (the relevant case when one surface is grounded or free), the full discontinuity gives:

$$E = \frac{D}{\epsilon_0 \epsilon_r} = \boxed{\frac{\sigma}{\epsilon_0 \epsilon_r}}$$

As $\sigma$ increases with continued irradiation, $E$ approaches the intrinsic dielectric strength ($E_b \approx 20\;\text{kV/mm}$ for PMMA). The system is metastable: the stored electrostatic energy density $u = \frac{1}{2}\epsilon_0\epsilon_r E^2$ grows quadratically with the field.

**Derivation of stored electrostatic energy.** The energy density in an electric field within a linear dielectric is:

$$u = \frac{1}{2}\vec{D} \cdot \vec{E} = \frac{1}{2}\epsilon_0\epsilon_r E^2$$

This follows from the work done assembling the charge distribution: bringing an incremental charge $dq$ to a region at potential $\phi$ costs energy $dU = \phi\,dq$, and integrating over the full distribution yields $U = \frac{1}{2}\int \rho\phi\,dV$, which can be rewritten via Gauss's law as $U = \frac{1}{2}\int \epsilon_0\epsilon_r E^2\,dV$.

For a uniform field $E = \sigma/(\epsilon_0\epsilon_r)$ throughout a slab of area $A$ and thickness $d$:

$$U = \frac{1}{2}\epsilon_0\epsilon_r E^2 \cdot (A d) = \frac{1}{2}\epsilon_0\epsilon_r \left(\frac{\sigma}{\epsilon_0\epsilon_r}\right)^2 A d = \boxed{\frac{\sigma^2 A d}{2\epsilon_0\epsilon_r}}$$

At representative parameters ($\sigma \sim 5\;\mu\text{C/cm}^2$, $A \sim 25\;\text{cm}^2$, $d \sim 2\;\text{cm}$, $\epsilon_r \sim 3$), $U \approx 47\;\text{J}$, released over nanosecond timescales — yielding instantaneous power dissipation on the order of gigawatts.

### 4.3 Breakdown Initiation

The bulk field is near-uniform and may remain marginally sub-critical; initiation requires a localized perturbation that drives the field past threshold. A conductive object (typically a steel nail) is driven into the specimen by mechanical impact, exploiting two complementary mechanisms:

**Mechanism 1 — Dielectric barrier reduction.**

*Derivation.* The nail punctures to within distance $d_{\text{eff}}$ of the space charge layer. The potential difference $\Delta V$ between the conductor (approximately at the surface potential) and the charge plane is fixed by the embedded charge. The field in the remaining dielectric gap is simply:

$$E_{\text{local}} = \frac{\Delta V}{d_{\text{eff}}}$$

This follows directly from $E = -d\phi/dz$ for a linear potential drop across a uniform gap. As $d_{\text{eff}} \to 0$:

$$\boxed{E_{\text{local}} \to \infty}$$

so breakdown is guaranteed regardless of the bulk field strength.

**Mechanism 2 — Geometric field enhancement.** The sharp nail tip ($r_{\text{tip}} \sim 10$–$100\;\mu\text{m}$) concentrates the field via $E_{\text{tip}} \propto 1/r_{\text{tip}}$ (derived in §2.3), with enhancement factors $\beta \sim 10^2$–$10^3$.

Both act simultaneously. The conductor need not be explicitly grounded — the nail provides a conductive path into which the trapped charge begins to redistribute, and the first plasma channel initiates the self-sustaining discharge.

### 4.4 Streamer Propagation

The discharge advances as a **streamer** — a filamentary ionization wave driven by the space-charge field at its head:

1. Free electrons at the channel tip are accelerated by the local field
2. At energies exceeding the ionization threshold ($\sim$10–15 eV for PMMA molecular bonds), inelastic collisions initiate an **electron avalanche** governed by the Townsend coefficient $\alpha(E)$
3. The avalanche generates a conducting plasma filament extending the channel — effectively advancing the grounded boundary into the charged volume
4. The new tip reconcentrates the field ($E_{\text{tip}} \propto r_{\text{tip}}^{-1}$), sustaining propagation
5. Stochastic asymmetries drive **tip-splitting bifurcations**, producing the dendritic morphology

Streamer velocities in solid dielectrics reach $10^5$–$10^7\;\text{m/s}$. The discharge converts stored energy into Joule heating, optical emission, acoustic shock, and irreversible fracture along the channel walls.

### 4.5 Characteristic Parameters

| Parameter | Symbol | Typical Value (PMMA) |
|---|---|---|
| Relative permittivity | $\epsilon_r$ | 2.6–3.5 |
| Intrinsic dielectric strength | $E_b$ | $\sim$20 kV/mm |
| Beam energy | $E_{\text{beam}}$ | 5–25 MeV |
| Surface charge density | $\sigma$ | 1–10 $\mu$C/cm$^2$ |
| Bulk resistivity | $\rho$ | $> 10^{15}\;\Omega\cdot$cm |
| Thermal conductivity | $k$ | 0.19 W/m$\cdot$K |
| Fractal dimension (2D projection) | $D_f$ | $\approx$ 1.7 |
| Discharge timescale | $\tau$ | $\sim$ns |

# Part III — Resistive Surface Burning (Wood Method)
---

## 5. Formation Process

The wood burning method produces Lichtenberg figures through a fundamentally different physical process than the acrylic method. Rather than impulsive dielectric breakdown of a bulk insulator, it relies on sustained current-driven Joule heating of a resistive surface layer, with a positive feedback loop between carbonization and conductivity.

### 5.1 Surface Preparation and Circuit

A wood substrate is wetted with an electrolyte solution — typically water with dissolved NaHCO$_3$ (baking soda) or NaCl — to create a thin, weakly conductive surface film. Two electrodes (nails, clamps, or wire contacts) are attached at separated points on the surface and connected to a high-voltage, high-current source.

The electrolyte concentration controls the initial surface conductivity $\sigma_{\text{cond}}$ and therefore the initial current distribution and power dissipation profile.

### 5.2 Current Flow and Joule Heating

When voltage is applied, current flows through the wetted surface layer between the electrodes.

**Derivation of Joule heating power density.** Start from the macroscopic relation $P = IV$. For a volume element of material with cross-sectional area $A$ and length $\ell$:

- Current: $I = JA$
- Voltage drop: $V = E\ell$

The power dissipated in the element is:

$$P = IV = (JA)(E\ell) = JE \cdot (A\ell)$$

Since $A\ell$ is the volume, the power per unit volume is:

$$p = \vec{J} \cdot \vec{E}$$

Now apply Ohm's law ($\vec{J} = \sigma_{\text{cond}}\vec{E}$) to eliminate one variable:

- Substitute $\vec{J}$: $\;p = (\sigma_{\text{cond}}\vec{E}) \cdot \vec{E} = \sigma_{\text{cond}}|\vec{E}|^2$
- Substitute $\vec{E} = \vec{J}/\sigma_{\text{cond}}$: $\;p = \vec{J} \cdot \vec{J}/\sigma_{\text{cond}} = |\vec{J}|^2/\sigma_{\text{cond}}$

Therefore:

$$\boxed{p = \sigma_{\text{cond}} |\vec{E}|^2 = \frac{|\vec{J}|^2}{\sigma_{\text{cond}}}}$$

This is the volumetric form of $P = I^2R$. The two forms highlight complementary physics: the first says more conductive material in a given field dissipates more power (more current flows); the second says for a given current density, more resistive material heats faster (more resistance does more work).

The current distribution is governed by the 2D Laplace equation for the surface potential (in the quasi-static limit), but the effective conductivity $\sigma_{\text{cond}}(\vec{r})$ is **spatially heterogeneous** due to the microstructure of wood:

- **Grain orientation**: Conductivity along the grain (longitudinal tracheids/vessels) can be 2–10$\times$ higher than across it, introducing strong anisotropy
- **Density variations**: Earlywood (low density, large cell lumens) vs. latewood (high density, thick cell walls) conduct differently
- **Moisture distribution**: Local variations in electrolyte penetration create conductivity gradients
- **Knots, resin pockets, and defects**: Act as barriers or preferential conduction paths

The current therefore does not flow uniformly but concentrates along low-resistance paths, producing localized hotspots where $p$ is highest.

### 5.3 Carbonization Feedback Loop

The critical mechanism that produces branching is a **positive feedback loop** between temperature and conductivity.

**Derivation of the heat equation with Joule source.** Conservation of energy for a volume element requires that the rate of temperature increase equals the net heat input:

$$\underbrace{\rho_m c_p \frac{\partial T}{\partial t}}_{\text{rate of thermal energy storage}} = \underbrace{p}_{\text{Joule heating source}} + \underbrace{k\nabla^2 T}_{\text{thermal conduction (in/out)}}$$

The Joule source term $p = |\vec{J}|^2/\sigma_{\text{cond}}$ was derived above. The conduction term comes from Fourier's law ($\vec{q} = -k\nabla T$) and conservation of energy ($\partial_t(\rho_m c_p T) = -\nabla \cdot \vec{q} + p$), giving:

$$\boxed{\rho_m c_p \frac{\partial T}{\partial t} = \frac{|\vec{J}|^2}{\sigma_{\text{cond}}} + k\nabla^2 T}$$

where $\rho_m$ is the mass density, $c_p$ the specific heat capacity, and $k$ the thermal conductivity.

The feedback loop proceeds as:

1. **Localized heating**: At hotspots where current density is highest, $p$ is largest and the temperature rises fastest according to the heat equation above.

2. **Pyrolysis and carbonization**: When the local temperature exceeds 200–300C, the cellulose and lignin matrix begins to undergo thermal decomposition (pyrolysis). Above 350C, this produces amorphous carbon (char), which is electrically conductive — orders of magnitude more so than intact wood.

3. **Conductivity enhancement**: The carbonized channel now carries substantially more current, increasing the local Joule heating further, advancing the carbonization front, and extending the conducting path.

4. **Branching**: At the advancing front of the carbonized region, the current must spread into intact (poorly conducting) wood. Where the wood microstructure offers competing paths of similar resistance, the current splits and the burn front bifurcates — producing the characteristic dendritic branching.

5. **Completion**: The branching fronts from both electrodes grow toward each other. When they meet, a low-resistance carbonized path bridges the electrodes, the current surges (often producing visible arcing), and the process terminates — sometimes violently.

### 5.4 Role of the Electrolyte

The electrolyte serves several functions:

- **Initial conductivity**: Dry wood is effectively insulating ($\sigma_{\text{cond}} < 10^{-8}\;\text{S/m}$). The electrolyte raises the surface conductivity to $\sim 10^{-2}$–$10^{0}\;\text{S/m}$, enabling current flow at accessible voltages.
- **Thermal mass and evaporation**: The water moderates the initial heating rate. As burning progresses, water evaporates from the hottest regions, leaving behind salt deposits that may further modify local conductivity.
- **Concentration dependence**: Higher electrolyte concentration reduces the overall resistance, increasing the total current and power dissipation but also producing more uniform current distributions (less branching). Lower concentrations yield higher contrast between conductive and insulative regions, promoting finer branching.

### 5.5 Characteristic Parameters

| Parameter | Symbol | Typical Value |
|---|---|---|
| Wood dielectric constant | $\epsilon_r$ | 2–6 (species- and moisture-dependent) |
| Dry wood resistivity | $\rho$ | $> 10^{8}\;\Omega\cdot$cm |
| Wetted surface conductivity | $\sigma_{\text{cond}}$ | $10^{-2}$–$10^{0}$ S/m |
| Carbonized wood resistivity | $\rho_{\text{char}}$ | $\sim 10^{-1}$–$10^{1}\;\Omega\cdot$cm |
| Pyrolysis onset temperature | $T_{\text{pyr}}$ | $\sim$200–300$^\circ$C |
| Char formation temperature | $T_{\text{char}}$ | $\sim$350$^\circ$C |
| Wood thermal conductivity | $k$ | 0.12–0.17 W/m$\cdot$K |
| Fractal dimension | $D_f$ | $\approx$ 1.3–1.6 |
| Process timescale | $\tau$ | seconds to minutes |

# Part IV — Comparative Analysis and Natural Manifestations
---

## 6. Comparison of Formation Methods

| Property | Acrylic (Charge-Implanted) | Wood (Resistive Burning) |
|---|---|---|
| **Physical mechanism** | Impulsive dielectric breakdown of bulk insulator | Sustained Joule heating and carbonization of resistive surface |
| **Energy source** | Stored electrostatic energy from implanted space charge | Continuous power delivery from external HV source |
| **Timescale** | Nanoseconds (single impulsive event) | Seconds to minutes (quasi-static process) |
| **Propagation** | Streamer: ionization-wave plasma channels | Thermal: advancing carbonization front |
| **Branching driver** | Laplacian field instability in homogeneous medium | Current concentration in heterogeneous, anisotropic medium |
| **Substrate role** | Passive insulator (homogeneous, isotropic) | Active participant (grain, density, moisture control morphology) |
| **Fractal dimension** | $D_f \approx 1.65$–$1.75$ | $D_f \approx 1.3$–$1.6$ |
| **DLA correspondence** | Strong (Laplacian growth in isotropic medium) | Weak (anisotropy and thermal effects break DLA universality) |
| **Dimensionality** | 3D internal structure (viewed through transparent medium) | 2D surface pattern |
| **Reversibility** | Permanent fracture networks in rigid polymer | Permanent carbonized channels in cellulose matrix |
| **Equipment** | Linear accelerator or Marx generator | HV transformer (MOT, NST) |
| **Polarity dependence** | Strong (cathode- vs. anode-directed streamers) | Weak (thermal mechanism is largely polarity-independent) |

### 6.1 Unified Description

Despite their mechanistic differences, both methods share the essential ingredients for fractal pattern formation:

1. **A scalar field** (electrostatic potential or thermal field) governed by a Laplacian or diffusive equation
2. **A moving boundary** (streamer front or carbonization front) that advances preferentially where the field gradient is largest
3. **A positive feedback mechanism** (field enhancement at conducting tips, or conductivity increase upon carbonization) that amplifies perturbations
4. **Stochastic heterogeneity** (material defects or grain structure) that seeds the initial perturbations leading to bifurcation

The acrylic method produces figures closer to the idealized DLA/Laplacian growth universality class because the medium is homogeneous and isotropic. Wood figures are shaped by the strong material anisotropy, placing them in a different (anisotropic, thermally-coupled) growth regime with characteristically lower fractal dimensions and morphology that is partially deterministic (grain-following) and partially stochastic.

## 7. Natural and Engineering Manifestations

The physics underlying Lichtenberg figure formation operates across many orders of magnitude in spatial scale. Each natural example maps onto one (or both) of the two laboratory methods:

- **Atmospheric lightning.** The stepped leader and return stroke constitute a Lichtenberg figure formed by dielectric breakdown of air at atmospheric pressure, with channel lengths of $\mathcal{O}(1\text{–}10\;\text{km})$. The branching arises from the same Laplacian growth instability as in the acrylic method; measured fractal dimensions fall in $D_f \approx 1.5$–$1.7$. Paschen's law governs the breakdown threshold, with the minimum for air at $pd \approx 0.57\;\text{Torr}\cdot\text{cm}$ ($V_{b,\min} \approx 327\;\text{V}$).

- **Keraunographic marks (lightning figures on skin).** Transient, fern-like erythematous patterns on lightning strike survivors. These arise from surface current flow along the resistive skin — mechanistically closer to the wood burning process than to bulk dielectric breakdown. The subcutaneous vasculature and sweat ducts provide preferential conduction paths analogous to wood grain.

- **Fulgurites.** Hollow glass tubes formed when lightning strikes silica-rich substrates (sand, soil, rock). The channel temperature ($\sim$30,000 K) exceeds the melting point of SiO$_2$ (1,713$^\circ$C), fusing the surrounding material. Fulgurite morphology preserves the 3D branching topology of sub-surface Laplacian growth in a heterogeneous medium.

- **Electrical treeing in power cable insulation.** In cross-linked polyethylene (XLPE) insulation of high-voltage transmission cables, sustained partial discharges at defect sites nucleate branching degradation channels. This is mechanistically identical to the acrylic method but occurs over timescales of months to decades. Understanding electrical treeing is a central problem in power systems reliability engineering.
