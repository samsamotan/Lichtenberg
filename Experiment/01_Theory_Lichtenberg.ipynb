# Theoretical Foundations of Lichtenberg Figure Formation

*Physics II — Computation Project (Theory Notebook)*

## Section 1: Physical Background

#### 1.1 Dielectric Breakdown

A *dielectric* is a material whose conduction band is, under ordinary conditions, inaccessible to the available charge carriers; current flow through the bulk is therefore negligible. When the electric field inside the dielectric grows past a material-dependent threshold $E_{\mathrm{bd}}$ — the *dielectric strength* — the insulator loses this property abruptly and becomes a conductor along narrow, filamentary channels. The process is called **dielectric breakdown** [1]. A partial list of mechanisms feeding into the transition includes direct field ionisation, impact ionisation in a Townsend-type electron avalanche, and in solids the formation of percolation paths through defect structure; a unified discussion is given in Ray [1] and, for gases specifically, in Meek and Craggs [2].

Once a breakdown channel forms, current flows through it and deposits Joule heat. In wood, which is the medium of interest for this project, the heat chars the cellulose along the channel, and carbonised wood is itself a moderate conductor. The conducting tip therefore advances through the insulator, and the visible burn is the locus of its trajectory.


#### 1.2 Lichtenberg Figures

The burn pattern is never a straight line between the two electrodes. It branches repeatedly and self-similarly, producing the dendritic, tree-like structures known as **Lichtenberg figures**, first recorded by G. C. Lichtenberg in 1777 and later classified and photographed systematically by Takahashi [3] and by Niemeyer and Pinnekamp [4]. The branching arises because the growing tip sits at a *singular* point of the electrostatic field: tips enhance the local field while concave regions (inside an already-carbonised pocket) are screened. The physics responsible for the geometry is therefore electrostatic, not chemical — the chemistry of wood only determines how visibly the trajectory is recorded.

For wood burning with two surface electrodes, two figures grow inward simultaneously, one from each electrode, until their tips meet. The present project targets this wood-on-surface geometry; the electrodynamic model described below was originally developed by Niemeyer, Pietronero, and Wiesmann [5] for planar surface discharges in $\mathrm{SF}_6$ gas, but the governing equation and the stochastic growth rule carry over verbatim.


#### 1.3 Scope

The theoretical object we need is therefore a rule that, given the current shape of the carbonised region, tells us where it is likely to grow next. Because the channels themselves are (to a very good approximation) equipotential and because we work on timescales long compared to the dielectric relaxation time but short compared to the growth timescale, that rule follows from the *electrostatics* of the carbonised region embedded in the insulator. The remainder of this notebook derives that rule and the scaling properties of the patterns it produces.

## Section 2: Electrostatics of the Breakdown Region

#### 2.1 Maxwell's Equations in the Quasi-Static Limit

In vacuum, Maxwell's equations in Gaussian form read

$$
\nabla \cdot \mathbf{E} = 4\pi \rho, \quad
\nabla \cdot \mathbf{B} = 0, \quad
\nabla \times \mathbf{E} = -\frac{1}{c}\frac{\partial \mathbf{B}}{\partial t}, \quad
\nabla \times \mathbf{B} = \frac{4\pi}{c}\mathbf{J} + \frac{1}{c}\frac{\partial \mathbf{E}}{\partial t}.
$$

Inside a linear dielectric we replace $\mathbf{E}$ by the displacement $\mathbf{D} = \varepsilon \mathbf{E}$ in the first equation and by $\mathbf{H} = \mathbf{B}/\mu$ in the fourth; we assume $\varepsilon$ and $\mu$ piecewise constant. For Lichtenberg-figure growth the relevant timescale is the channel-propagation time $\tau_{\mathrm{prop}} \sim 10^{-6}\,\mathrm{s}$, which is long compared to the dielectric relaxation time of wet wood ($\tau_{\mathrm{rel}} = \varepsilon/\sigma \sim 10^{-9}\,\mathrm{s}$) but short compared to the thermal timescale. Under these conditions the magnetic terms are subdominant by a factor $(\ell/c\tau_{\mathrm{prop}})^2 \ll 1$, and the field is **quasi-static**:

$$
\nabla \times \mathbf{E} \approx 0.
$$

This is the regime used by Niemeyer *et al.* [5] and by every subsequent DBM paper we are aware of.


#### 2.2 Scalar Potential and Poisson's Equation

Because $\mathbf{E}$ is curl-free it admits a scalar potential $\phi$ with

$$
\mathbf{E} = -\nabla \phi.
$$

Combining this with Gauss's law in the dielectric,

$$
\nabla \cdot (\varepsilon \nabla \phi) = -4\pi \rho_{\mathrm{f}},
$$

and assuming $\varepsilon$ constant inside the insulator gives **Poisson's equation**,

$$
\nabla^{2} \phi = -\frac{4\pi}{\varepsilon}\,\rho_{\mathrm{f}}.
$$

Between the electrodes and outside the carbonised region, the wood carries no free charge at the moment of interest (the charges deposited during breakdown have already relaxed to the channel boundary), so $\rho_{\mathrm{f}} = 0$ and Poisson's equation reduces to **Laplace's equation**:

$$
\boxed{\;\nabla^{2} \phi = 0\;}
$$

This is the single equation on which the rest of the project is built.


#### 2.3 The Channel as an Equipotential

Carbonised wood has a conductivity many orders of magnitude larger than virgin wood. As long as the total current the external supply can push through the channel is small enough that the ohmic drop along it is negligible compared to the applied voltage, we may treat the entire carbonised region as a single equipotential $\phi = \phi_{\mathrm{ch}}$. Without loss of generality we fix $\phi_{\mathrm{ch}} = 0$; the other electrode is then fixed at some non-zero value set by the transformer.

This reduction matters computationally because it collapses the geometry of the partially-burned region to a *Dirichlet boundary condition*: at every lattice site that is already part of the tree we impose $\phi = 0$; at lattice sites on the supply electrode we impose $\phi = \phi_{0}$; elsewhere $\phi$ is governed by $\nabla^{2}\phi = 0$. Niemeyer *et al.* [5] call this the *equipotential channel approximation* and Pietronero and Wiesmann [6] argue that it is accurate for all gaseous and most solid dielectrics.


#### 2.4 Boundary Conditions for the Wood Geometry

The boards used in fractal wood-burning videos are rectangular, with two probes placed on opposite short edges. The chosen numerical domain is therefore an $m \times n$ rectangle with:

* Top edge (cathode): $\phi = -V_{0}$.
* Bottom edge (anode): $\phi = +V_{0}$.
* Left and right edges: periodic, to eliminate spurious side-wall screening in a plate much wider than the electrode gap.

The symmetric choice $\pm V_{0}$ rather than $0$ and $V_{0}$ puts the neutral surface in the middle of the board and gives the two growing trees a symmetric attractor, which is required for their geometries to come out statistically equivalent. This is the reason the sign convention differs from the single-tree (lightning) case, where the natural choice is $\phi = 0$ at the seed and $\phi = 1$ at the far boundary.

## Section 3: The Dielectric Breakdown Model

#### 3.1 Motivation

Solving Laplace's equation tells us the field everywhere, but it does not by itself tell us *where the channel will advance next*. A deterministic rule — "grow always in the direction of the largest field" — produces a single straight line, which is not what experiments show. The actual process is stochastic: the avalanches that initiate each advance are born from individual seed electrons whose appearance is a Poisson event [2]. The stochastic element must therefore enter the growth rule.

Niemeyer, Pietronero, and Wiesmann [5] proposed the minimal such rule: the probability that a given perimeter site is the *next* site added to the cluster is a power of the local potential at that site, normalised over the perimeter. We now derive the form of that power law and spell out the rule in full.


#### 3.2 Growth Probability

Let $\Omega$ denote the current (discrete) breakdown channel, and let $\partial\Omega$ denote the set of lattice sites not in $\Omega$ but sharing an edge with at least one site in $\Omega$. For each candidate site $s \in \partial\Omega$, let $\phi_{s}$ denote the potential at $s$, computed from the Laplace equation with boundary conditions $\phi = 0$ on $\Omega$ and the external $\phi = V_{0}$ on the supply electrode.

Niemeyer *et al.* [5] assume that the growth probability at site $s$ is

$$
p_{s} \;=\; \frac{\phi_{s}^{\,\eta}}{\displaystyle\sum_{s' \in \partial\Omega} \phi_{s'}^{\,\eta}}, \qquad \eta \ge 0.
$$

The exponent $\eta$ encodes the unknown microscopic relation between local field and avalanche rate. Because $|\mathbf{E}| \propto |\nabla\phi|$ and $\phi_{\mathrm{ch}} = 0$, for bonds one lattice step long the local field magnitude and the potential at the candidate site are proportional, so $p_{s}$ is equivalently $\propto |\mathbf{E}_{s}|^{\,\eta}$.


#### 3.3 Interpretation of the Exponent

$\eta$ interpolates between two physically distinguished limits.

For $\eta = 0$ every perimeter site is equally likely; the field plays no role and the model degenerates to the **Eden model** of random cluster growth [7]. Eden clusters are compact: their fractal dimension equals the embedding dimension, $D = 2$ in the plane.

For $\eta \to \infty$ only the site with the largest $\phi$ is ever chosen; the stochasticity vanishes and the cluster grows along whichever tip has the locally-largest field. The structure contracts to a one-dimensional filament, $D \to 1$.

For intermediate $\eta$ the competition between these two limits produces self-similar, multiply-branched structures. Niemeyer *et al.* [5] report $D = 1.75 \pm 0.02$ at $\eta = 1$ on a two-dimensional square lattice with radial geometry, in agreement with the value $D \simeq 1.7$ extracted from their own photographs of planar surface discharges in compressed $\mathrm{SF}_{6}$. Kim *et al.* [8] later mapped out the entire $\eta$ axis from $0$ to $4$ and confirmed that $D$ decreases monotonically from $2$ to $1$ as $\eta$ grows.

The physical value of $\eta$ is empirical. For gases Niemeyer *et al.* argue $\eta = 1$ gives the correct dimension; for solids and polymers with quenched disorder, Wiesmann and Zeller [9] and Pietronero and Wiesmann [6] argue that a nonlinear dependence $\eta \neq 1$ is more realistic. For wood we do not know $\eta$ a priori, and part of the computational exercise is to sweep it.


#### 3.4 The Growth Process

The full evolution is now a Markov chain on the set of subsets of the lattice:

1. Start from a seed, typically a single site on the supply electrode.
2. Solve $\nabla^{2}\phi = 0$ with the current boundary conditions.
3. Enumerate the perimeter $\partial\Omega$ and form the weights $\phi_{s}^{\eta}$.
4. Sample a single site from $\partial\Omega$ with probability $p_{s}$ above and add it to $\Omega$. Fix $\phi = 0$ at the new site.
5. Return to step 2, unless the cluster has touched the opposing electrode.

Step 2 is the numerically expensive one and is deferred to the computation notebook. Note that steps 2–5 are run *alternately* for the two trees in a two-electrode setup: each tree contributes a new bond before the field is re-solved and the other tree contributes its next bond. This alternation keeps the two trees statistically equivalent.


#### 3.5 Relation to Diffusion-Limited Aggregation

The DBM at $\eta = 1$ is mathematically very close to **diffusion-limited aggregation** (DLA) of Witten and Sander [10]. In DLA, particles released one at a time at infinity perform random walks and stick to the cluster on first contact. The probability that a random walk first contacts the cluster at a given perimeter site is the *harmonic measure* — the same object that gives the local field at that site when the cluster is held at $\phi = 0$ and infinity at $\phi = 1$ [11]. DBM at $\eta = 1$ therefore samples the harmonic measure directly, and both models predict $D \approx 1.7$ in two dimensions (DLA: $D \simeq 1.71$ [10]; DBM: $D \simeq 1.75$ [5]).

The difference lies elsewhere: for $\eta \neq 1$ DBM generates a continuous family of patterns parametrised by $\eta$, whereas DLA is a single point in that family. The ability to tune $\eta$ is the reason DBM, rather than DLA, is the appropriate model for a material-specific phenomenon such as Lichtenberg burning.

## Section 4: Fractal Geometry of the Discharge Pattern

#### 4.1 Self-Similarity and the Hausdorff Dimension

The patterns produced by DBM are *fractal* in the sense of Mandelbrot [12]: their scaling is described by a non-integer **Hausdorff dimension** $D$. For a sufficiently regular set $S \subset \mathbb{R}^{2}$, $D$ may be characterised operationally through the **box-counting** or **Minkowski–Bouligand** dimension:

$$
D \;=\; \lim_{\varepsilon \to 0} \frac{\log N(\varepsilon)}{\log(1/\varepsilon)},
$$

where $N(\varepsilon)$ is the minimum number of boxes of side $\varepsilon$ needed to cover $S$. For self-similar sets of reasonable regularity the box-counting dimension coincides with the Hausdorff dimension [13], and throughout this project we use them interchangeably, writing simply $D$.


#### 4.2 Radial Scaling

For a DBM cluster grown from a point electrode embedded in a disk, Niemeyer *et al.* [5] characterise $D$ through an equivalent *radial* law. Let $N(r)$ denote the number of cluster sites inside a circle of radius $r$ centred on the seed. If the cluster is a fractal of dimension $D$, then

$$
N(r) \;\sim\; r^{\,D},
$$

and the branch density at distance $r$ is

$$
n(r) \;\sim\; \frac{dN}{dr} \;\sim\; r^{\,D-1}.
$$

Niemeyer *et al.* extract $D \simeq 1.7$ from photographs by counting branches at successive radii; we will reproduce the same measurement numerically in the computation notebook, after cross-validating the box-counting routine on the Sierpiński carpet, whose $D = \log_{3} 8 \approx 1.893$ is known exactly [12].


#### 4.3 Dependence on $\eta$

Combining the two limits from §3.3 with the intermediate simulation at $\eta = 1$, Niemeyer *et al.* [5] obtain Table I below (reproduced from their paper). The $\eta = 2$ value is cited from an earlier reference of Sawada *et al.* [14] and is attached an informal uncertainty.

| $\eta$ | $D$ |
| :---: | :---: |
| $0$ | $2$ |
| $0.5$ | $1.89 \pm 0.01$ |
| $1$ | $1.75 \pm 0.02$ |
| $2$ | $\sim 1.6$ |

No universality class exists: $D$ depends on $\eta$, which in turn depends on the medium. The non-universality is physically important because it means that *what Lichtenberg figures look like is material-specific*. Figures in gases are not the same object as figures in oil or in wood, even though the same Laplacian skeleton underlies all three.


#### 4.4 Debye Screening

A further refinement of the model is relevant to our geometry. Real wood is not a perfect insulator: moisture and the applied electrolyte contribute mobile charge carriers, which screen fields beyond a length

$$
\lambda_{\mathrm{D}} \;=\; \sqrt{\frac{\varepsilon k_{\mathrm{B}} T}{4\pi\sum_{i} n_{i} q_{i}^{2}}}.
$$

Beyond $\lambda_{\mathrm{D}}$ the Laplace equation is modified to the *Helmholtz* (screened-Poisson) equation $(\nabla^{2} - \lambda_{\mathrm{D}}^{-2})\phi = 0$. Niemeyer *et al.* [5] note that screening breaks self-similarity: below $\lambda_{\mathrm{D}}$ the pattern is still fractal, but above it the growth becomes compact with $D \to 2$. We neglect screening in the present work because the electrode gap in the wood-burning experiment is small compared to $\lambda_{\mathrm{D}}$ at the applied voltages of interest [4], but we flag it here for completeness.

## Section 5: Anisotropy

#### 5.1 Why Wood is Different from $\mathrm{SF}_{6}$

Niemeyer *et al.* [5] study a homogeneous, isotropic gas. Wood is neither. Its structure is that of a bundle of longitudinal fibres (tracheids and vessels) with high aspect ratio, oriented by the growth direction of the tree; the conductivity is consequently several times larger along the grain than across it. To first approximation, this is a *quenched* disorder: the random variation from point to point of the material's response is frozen in at the time the tree grew, and does not evolve during a single burn.

Pietronero and Wiesmann [6] analysed DBM with quenched disorder and showed that the presence of a small random component in the growth probability does not, in two dimensions, change the fractal dimension $D$. It does, however, align branches preferentially along the directions of lower resistance, producing the visually characteristic elongation of Lichtenberg figures in wood along the grain.


#### 5.2 Implementation Scheme

The cleanest way to introduce anisotropy into the DBM is to multiply the bare growth weights $\phi_{s}^{\,\eta}$ by a site-dependent *conductance* $g_{s} > 0$:

$$
p_{s} \;\propto\; g_{s}\, \phi_{s}^{\,\eta}.
$$

For isotropic wood, $g_{s}$ is drawn i.i.d. from a distribution reflecting microstructural inhomogeneity — an exponential or log-normal distribution with a small coefficient of variation is adequate [6]. For anisotropic wood one replaces $g_{s}$ with a direction-dependent weight that enhances transitions along the grain; we implement this in the computation notebook. Pietronero and Wiesmann report that the statistics of $D$ are robust to either choice, so we take the simpler binary Bernoulli randomisation of Chen and Gao [15] for the production runs and verify agreement with the log-normal form as a sanity check.

## Section 6: Summary and Predictions

We have reduced the physical problem of Lichtenberg figure formation on wood to the following mathematical skeleton.

The potential $\phi$ satisfies Laplace's equation $\nabla^{2}\phi = 0$ on the part of the board that is not yet burned, with $\phi = 0$ on the burned region $\Omega$ and $\phi = \pm V_{0}$ on the two electrodes. The burned region grows by a stochastic Markov rule: at each step, a single new perimeter site is added with probability $p_{s} \propto \phi_{s}^{\,\eta}$, after which the Laplace equation is re-solved. Two such processes run in parallel, one from each electrode, with potentials chosen so that the two are symmetric. Anisotropy of the medium enters as a site-dependent conductance prefactor.

Three concrete physical predictions follow from this skeleton and will be tested numerically:

1. The resulting patterns are **fractal**, with a well-defined box-counting dimension $D$.
2. $D$ **decreases monotonically** from $D = 2$ (Eden, compact) at $\eta = 0$ towards $D = 1$ (straight line) as $\eta \to \infty$, crossing $D \simeq 1.75$ at the physically-motivated value $\eta = 1$ for isotropic planar geometry [5].
3. Two trees grown simultaneously from opposite electrodes with symmetric boundary conditions produce approximately symmetric figures and meet near the midline of the board; the attraction between them is a direct consequence of the boundary conditions, not a hand-coded ingredient.

The first two of these predictions are quantitative and allow the simulation to be falsified. The third is qualitative but sharp enough to distinguish DBM-grown patterns from the output of the breadth-first-search "least-resistance" heuristic that was used in the earlier, superseded version of this project [15].
